# Phi-3

The first step is to load our model onto the GPU for faster inference. Note that we load the model and tokenizer separately (although that isn't always necessary).

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# Define the model ID
model_id = "gpt2"

# Load model and tokenizer
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="cpu",
    torch_dtype=torch.float32
)
tokenizer = AutoTokenizer.from_pretrained(model_id)

# Set pad token to eos token to avoid padding warnings
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
```

Although we can now use the model and tokenizer directly, it's much easier to wrap it in a `pipeline` object:

```python
from transformers import pipeline

# Create a pipeline
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    return_full_text=False,
    max_new_tokens=500,
    do_sample=False
)
```

Finally, we create our prompt as a user and give it to the model:

```python
# The prompt for the base model
prompt = "Create a funny joke about chickens."

# Tokenize the input and move to the correct device
input_ids = tokenizer.encode(prompt, return_tensors="pt").to(model.device)

# Generate output directly from the model
outputs = model.generate(
    input_ids,
    max_new_tokens=50,
    pad_token_id=tokenizer.eos_token_id
)

# Decode the output
response_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(response_text)
```
